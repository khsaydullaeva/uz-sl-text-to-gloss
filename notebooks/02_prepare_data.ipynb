```python
import sys
from pathlib import Path

ROOT = Path("..").resolve()
sys.path.insert(0, str(ROOT / "src"))

from ttg.config import DATA_DIR, CHECKPOINTS_DIR, ARTIFACTS, PROJECT_ROOT
print("PROJECT_ROOT =", PROJECT_ROOT)
print("DATA_DIR     =", DATA_DIR)
print("CHECKPOINTS  =", CHECKPOINTS_DIR)
```

```python
DEV_RATIO = 0.1
TEST_RATIO = 0.1
SEED = 42
LOWERCASE_GLOSS = True
```

```python
import json
from ttg.data import latest_raw_dataset, prepare_splits

input_path = latest_raw_dataset()
print("Using latest raw dataset:", input_path)

meta = prepare_splits(
    input_path=input_path,
    out_dir=DATA_DIR,
    dev_ratio=DEV_RATIO,
    test_ratio=TEST_RATIO,
    seed=SEED,
    lowercase_gloss=LOWERCASE_GLOSS,
)
print(json.dumps(meta, indent=2, ensure_ascii=False))
```
